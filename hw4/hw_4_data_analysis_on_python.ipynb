## Imports

```python
import random
import typing as tp
import numpy as np

from collections import Counter

random.seed(42, version=2)  # задаем зерно датчика случайных чисел
```

## Задача 1 (2.48)

```python
def task1(n_exp: int = 10 ** 3, k: int = 5, num_equal: int = 2):
    A = list(range(1, 7))  # пространство элементарных событий

    count = 0
    for i in range(1, n_exp + 1):
        # k бросков кубиков (выборка с возвращением)
        dice = random.choices(A, k=k)

        counter = Counter(dice)

        if set(counter.values()) & set(range(num_equal, k + 1)):
            count += 1

    return count / n_exp
```

Бросаются 5 игральных костей. Найти вероятность того, что по меньшей мере на двух (2) из них выпадут одинаковые грани.

**Аналитическое решение:**

N_total (всего вариантов выбросить кости) = $6^5$

N (всего вариантов выбросить кости так, чтобы все грани были различны) = $6 * 5 * 4 * 3 * 2$

P = $\frac{N}{N_{total}}$ = $\frac{6 * 5 * 4 * 3 * 2}{6^5} = \frac{49}{54} \approx 0.9074$

```python
task_1_1_answer = 49 / 54
task_1_1_monte_carlo = task1(10 ** 8, k=5, num_equal=2)
print(f"Аналитический ответ: {task_1_1_answer}")
print(f"Ответ с помощью Монте-Карло: {task_1_1_monte_carlo}")
print(f'От: {abs(task_1_1_answer - task_1_1_monte_carlo) / task_1_1_answer * 100:.3f}%')
```

Бросаются 5 игральных костей. Найти вероятность того, что по меньшей мере на трех (3) из них выпадут одинаковые грани.

**Аналитическое решение:**

N_total (всего вариантов выбросить кости) = $6^5$

N_5 (вариантов выбросить 5 одинаковых граней) = 6

N_4 (вариантов выбросить 4 одинаковых грани) = $6 * \binom{5}{4} * 5 = 6 * 5 * 5 = 150$

N_3 (вариантов выбросить 3 одинаковых грани) = $6 * \binom{5}{3} * 5 * 5 = 6 * 10 * 25 = 1500$

$P = \frac{N_5 + N_4 + N_3}{N_{total}} = \frac{6 + 150 + 1500}{6^5} = \frac{23}{108} \approx 0.21296$

```python
task_1_2_answer = 23 / 108
task_1_2_monte_carlo = task1(10 ** 8, k=5, num_equal=3)
print(f"Аналитический ответ: {task_1_2_answer}")
print(f"Ответ с помощью Монте-Карло: {task_1_2_monte_carlo}")
print(f'От: {abs(task_1_2_answer - task_1_2_monte_carlo) / task_1_2_answer * 100:.3f}%')
```

## Задача 2 (2.14)

```python
def task2_a(check: tp.Callable[[Counter["str"]], bool], n_exp: int = 10 ** 5, k: int = 3) -> float:
    #   Колода карт, нам важно только достоинство карты
    deck = []

    for i in range(4):
        deck.append("ace")
        deck.append("king")
        deck.append("queen")
        deck.append("jack")
        for j in range(2, 11):
            deck.append(str(j))

    count = 0
    for i in range(1, n_exp + 1):
        current_sample = random.sample(deck, k=k)

        current_counter = Counter(current_sample)

        if check(current_counter):
            count += 1
        
    return count / n_exp
```

Из колоды карт (52 листа) наудачу вынимаются три карты. Найти вероятность того, что среди них окажется ровно один туз.

**Аналитическое решение:**

N_1_ace (кол-во способов выбрать один туз) = $\binom{4}{1} = 4$

N_2_no_ace (кол-во способов выбрать 2 не туза) = $\binom{48}{2} = 1128$

N_total (общее кол-во способов выбрать 3 карты) = $\binom{52}{3} = 22100$

P = $\frac{N_{1_{ace}} * N_{2_{\text{no ace}}}}{N_{total}} = \frac{4 * 1128}{22100} = \frac{1128}{5525}$

```python
def only_one_ace(counter: Counter["str"]) -> bool:
    return (counter["ace"] == 1)

task_2_1_answer = 1128 / 5525
task_2_1_monte_carlo = task2_a(only_one_ace, 10 ** 8, k=3)
print(f"Аналитический ответ: {task_2_1_answer}")
print(f"Ответ с помощью Монте-Карло: {task_2_1_monte_carlo}")
print(f'От: {abs(task_2_1_answer - task_2_1_monte_carlo) / task_2_1_answer * 100:.3f}%')
```

Из колоды карт (52 листа) наудачу вынимаются три карты. Найти вероятность того, среди них окажется хотя бы один туз.

**Аналитическое решение:**

N_no_aces (кол-во способов выбрать 3 карты не туза) = $\binom{48}{3} = 17296$

N_total (общее кол-во способов выбрать 3 карты) = $\binom{52}{3} = 22100$

P = 1 - $\frac{N_{\text{no aces}}}{N_{total}} = 1 - \frac{17296}{22100} = \frac{1201}{5525}$

```python
def at_least_one_ace(counter: Counter["str"]) -> bool:
    return (counter["ace"] >= 1)

task_2_2_answer = 1201 / 5525
task_2_2_monte_carlo = task2_a(at_least_one_ace, 10 ** 8, k=3)
print(f"Аналитический ответ: {task_2_2_answer}")
print(f"Ответ с помощью Монте-Карло: {task_2_2_monte_carlo}")
print(f'От: {abs(task_2_2_answer - task_2_2_monte_carlo) / task_2_2_answer * 100:.3f}%')
```

Из колоды карт (52 листа) наудачу вынимаются три карты. Найти вероятность того, это будут тройка, семерка и туз (в любом порядке).

**Аналитическое решение:**

N_total (общее кол-во способов выбрать 3 карты) = $\binom{52}{3} = 22100$

N_1_ace (кол-во способов выбрать один туз) = $\binom{4}{1} = 4$

N_1_three (кол-во способов выбрать одну тройку) = $\binom{4}{1} = 4$

N_1_seven (кол-во способов выбрать одну семерку) = $\binom{4}{1} = 4$

P = $\frac{N_{1_{ace}} * N_{1_{three}} * N_{1_{seven}}}{N_{total}} = \frac{4 * 4 * 4}{22100} = \frac{16}{5525}$

```python
def exact_combination(counter: Counter["str"]) -> bool:
    return counter == Counter(["3", "7", "ace"])

task_2_3_answer = 16 / 5525
task_2_3_monte_carlo = task2_a(exact_combination, 10 ** 8, k=3)
print(f"Аналитический ответ: {task_2_3_answer}")
print(f"Ответ с помощью Монте-Карло: {task_2_3_monte_carlo}")
print(f'От: {abs(task_2_3_answer - task_2_3_monte_carlo) / task_2_3_answer * 100:.3f}%')
```

## Задача 3

Случайная точка A наудачу выбирается в прямоугольнике со сторонами 1 и 2. Найти вероятности событий: расстояние от A
а) до ближайшей стороны прямоугольника не превосходит x;

Аналитический ответ: $P(A) = 3x - 2x^2, \quad 0 < x \le \frac{1}{2}$

```python
def monte_carlo_border_proximity_probability(n_exp: int = 10 ** 7, rect_width: int = 1, rect_len: int = 2, x: float = 0.2) -> float:
    count = 0
    for i in range(1, n_exp + 1):
        point_x = np.random.uniform(high=rect_width)
        point_y = np.random.uniform(high=rect_len)

        if min(min(rect_width - point_x, point_x), min(rect_len - point_y, point_y)) <= x:
            count += 1
            
    return count / n_exp
```

```python
x = 0.2

task_3_1_answer = 3 * x - 2 * x ** 2
task_3_1_monte_carlo = monte_carlo_border_proximity_probability(x=x)

print(f"Аналитический ответ: {task_3_1_answer}")
print(f"Ответ с помощью Монте-Карло: {task_3_1_monte_carlo}")
print(f'От: {abs(task_3_1_answer - task_3_1_monte_carlo) / task_3_1_answer * 100:.3f}%')
```

б) до каждой стороны прямоугольника не превосходит x;

Аналитический ответ: $P(B) = 2x^2 - 3x + 1, \quad 0 < x < \frac{1}{2}$

```python
def monte_carlo_corner_zone_probability(n_exp: int = 10 ** 7, rect_width: int = 1, rect_len: int = 2, x: float = 0.2) -> float:
    count = 0
    for i in range(1, n_exp + 1):
        point_x = np.random.uniform(high=rect_width)
        point_y = np.random.uniform(high=rect_len)

        if point_x >= x and point_x <= (rect_width - x) and \
           point_y >= x and point_y <= (rect_len - x):
            count += 1
            
    return count / n_exp
```

```python
task_3_2_answer = 2 * x ** 2 - 3 * x + 1
task_3_2_monte_carlo = monte_carlo_corner_zone_probability(x=x)

print(f"Аналитический ответ: {task_3_2_answer}")
print(f"Ответ с помощью Монте-Карло: {task_3_2_monte_carlo}")
print(f'От: {abs(task_3_2_answer - task_3_2_monte_carlo) / task_3_2_answer * 100:.3f}%')
```

в) до каждой диагонали прямоугольника не превосходит x.

Аналитический ответ: $P(C) = \frac{5}{2} x^2 \quad \text{или} \quad P(C) = 2.5x^2, \quad 0 < x \le \frac{1}{\sqrt{5}}$

```python
def monte_carlo_diagonal_proximity_probability(n_exp: int = 10 ** 7, rect_width: int = 1, rect_len: int = 2, x: float = 0.5) -> float:
    count = 0
    for i in range(1, n_exp + 1):
        point_x = np.random.uniform(high=rect_width)
        point_y = np.random.uniform(high=rect_len)

        diagonal_length = np.sqrt(rect_width ** 2 + rect_len ** 2)
        dist_to_left_up_corner = np.sqrt(point_x ** 2 + (rect_len - point_y) ** 2)
        dist_to_right_down_corner = np.sqrt((rect_width - point_x) ** 2 + point_y ** 2)

        first_half_perimeter = (diagonal_length + dist_to_left_up_corner + dist_to_right_down_corner) / 2
        
        dist_to_first_diagonal = (2 / diagonal_length) * np.sqrt(first_half_perimeter * 
            (first_half_perimeter - diagonal_length) * (first_half_perimeter - dist_to_left_up_corner) * 
            (first_half_perimeter - dist_to_right_down_corner)
        )

        dist_to_left_down_corner = np.sqrt(point_x ** 2 + point_y ** 2)
        dist_to_right_up_corner = np.sqrt((rect_width - point_x) ** 2 + (rect_len - point_y) ** 2)

        second_half_perimeter = (diagonal_length + dist_to_left_down_corner + dist_to_right_up_corner) / 2
        
        dist_to_sec_diagonal = (2 / diagonal_length) * np.sqrt(second_half_perimeter *
            (second_half_perimeter - diagonal_length) * (second_half_perimeter - dist_to_left_down_corner) *
            (second_half_perimeter - dist_to_right_up_corner)
        )

        if max(dist_to_first_diagonal, dist_to_sec_diagonal) <= x:
            count += 1
            
    return count / n_exp
```

```python
task_3_3_answer = 5 * x ** 2 / 2
task_3_3_monte_carlo = monte_carlo_diagonal_proximity_probability(x=x)

print(f"Аналитический ответ: {task_3_3_answer}")
print(f"Ответ с помощью Монте-Карло: {task_3_3_monte_carlo}")
print(f'От: {abs(task_3_3_answer - task_3_3_monte_carlo) / task_3_3_answer * 100:.3f}%')
```
