# Лабораторная номер 3 по анализу данных на Python

## Импорты 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn import metrics
```

## Загружаем данные

```python
# Загружаем данные с помощью pandas метода read_csv
df = pd.read_csv("data/Econom_Cities_data.csv", sep=";")

# Смотрим на данные 
df.head(5)
```

```python
# типы данных для каждого столбца
# только Work имеет числовой тип int64
df.dtypes
```

```python
assert len(df["City"].unique()) == len(df)  # Все города уникальны
```

Так как города уникальны, можно установить колонку City в качестве индекса нашего pandas датафрейма.

```python
# Устанавливаем индексом колонку "City"
df = df.set_index("City")
```

## Предобработка данных

Преобразуем колонки Price и Salary, чтобы их тип был float64.

```python
df["Price"] = (df["Price"].apply(lambda x: x.replace(",", "."))).astype(float)
df["Salary"] = (df["Salary"].apply(lambda x: x.replace(",", "."))).astype(float)
```

```python
# типы данных для всех колонок
df.dtypes
```

Также, так как из первой лабораторной мы знаем, что Cairo и Jakarta — это выбросы, мы удаляем их из нашего датафрейма.

```python
df.loc[["Cairo", "Jakarta"]]
```

```python
# Убираем из нашего pandas датафрейма Каир и Джакарту
df = df.loc[~df.index.isin(["Cairo", "Jakarta"])]
```

Так как наши признаки имеют разный масштаб (например, значения признака Work по наблюдениям сильно больше значений признака Salary), нам надо стандартизировать их.

Для этого мы импортируем StandardScaler из sklearn и стандартизуем данные. 

```python
# Для стандартизации используем StandardScaler из scikit-learn
standard_scaler = StandardScaler()

# Записываем стандартизованные данные тоже в pandas.DataFrame с тем же индексом и теми же названиями колонок
df_scaled = pd.DataFrame(standard_scaler.fit_transform(df.values), index=df.index, columns=df.columns)
```

## Применяем метод DBSCAN

Попробуем применить DBSCAN с дефолтными параметрами

```python
# DBSCAN с дефолтными параметрами на наших стандартизированных данных
default_dbscan = DBSCAN()
default_dbscan.fit(df_scaled)

default_dbscan_labels = default_dbscan.labels_

df["cluster"] = default_dbscan_labels

# Смотрим солько элементов в каждом кластере, кластер -1 это выбросы
df["cluster"].value_counts()
```

Видим, что дефолтный DBSCAN плохо кластеризует наши данные - он определил 36 наблюдений как выбросы, а остальные 10 засунул в нулевой кластер.

Попробую сам подобрать значение eps и min_samples для DBSCAN

```python
dbscan = DBSCAN(eps=1.0, metric="euclidean", min_samples=2)
dbscan.fit(df_scaled)
df["cluster"] = dbscan.labels_
df["cluster"].value_counts()
```

Получилось не очень хорошо, большая часть наблюдений пошла в один кластер, а в остальных кластерах не более 2 наблюдений, плюс есть 4 выброса.

Попробую подобрать другие параметры (взять eps поменьше)

```python
dbscan = DBSCAN(eps=0.4, metric="euclidean", min_samples=2)
dbscan.fit(df_scaled)
df["cluster"] = dbscan.labels_
df["cluster"].value_counts()
```

Получилось также довольно плохо, 22 наблюдения в выбросах и еще 8 кластеров с разным количеством элементов (не более 5).

Попробую подобрать параметры с помощью gridsearch и метрики silhouette, найти лучшие параметры для алгоритма DBSCAN для метрики silhouette (она плохо работает с ленточными кластерами, но из первой лабораторной мы знаем, что в данных, с которыми мы работаем, не нужно искать ленточные кластеры).

```python
# Возможные eps ('радиус' epsilon окрестности)
eps_lst = np.linspace(start=0.05, stop=3.0, num=200)
# Возможные min_samples (кол-во соседей в epsilon окрестности для того чтобы быть корневой точкой)
min_samples_lst = range(1, 11)

# Лучшие значения silhouette score, лучший dbscan алгоритм
best_silhouette_score, best_dbscan = -np.inf, None

# Лучшее значение параметра eps и лучшее значение параметра min_samples
best_eps, best_min_samples = None, None

# Идем по сетке возможных eps и min_samples, лучшим dbscan считаем тот,
# у которого самый большой silhouette_score
for current_eps in eps_lst:
    for current_min_samples in min_samples_lst:
        current_dbscan = DBSCAN(eps=current_eps, metric="euclidean", min_samples=current_min_samples)

        current_dbscan.fit(df_scaled)

        current_labels = current_dbscan.labels_

        # Число кластеров, после отбрасывания выбросов.
        n_clusters_ = len(set(current_labels)) - (1 if -1 in current_labels else 0)

        if n_clusters_ > 1 and n_clusters_ < len(current_labels):
            current_silhouette_score = metrics.silhouette_score(df_scaled, current_labels)
            if current_silhouette_score > best_silhouette_score:
                best_silhouette_score = current_silhouette_score
                best_dbscan = current_dbscan
                best_eps = current_eps
                best_min_samples = current_min_samples
```

```python
print(f"Лучший silhouette_score равен {best_silhouette_score}\n")
print(f"eps ('радиус' epsilon окрестности) который привел к лучшему значению silhouette score равен {best_eps}\n")
print(f"min_samples (кол-во соседей в epsilon окрестности для того чтобы быть корневой точкой),\
который привел к лучшему значению silhouette score равен {best_min_samples}")
```

```python
# Создаем колонку cluster с кодом кластера для каждого наблюдения
# код -1 идет для выбросов
df["cluster"] = best_dbscan.labels_

# Смотрим сколько наблюдений (городов) в каждом кластере
df["cluster"].value_counts()
```

Используя лучшие параметры с точки зрения метрики silhouette, у нас получилось кластеризовать данные на 4 кластера + кластер выбросов.

```python
# Средние значения признаков в каждом кластере
df.groupby("cluster").mean()
```
